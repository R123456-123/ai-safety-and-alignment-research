```python
import asyncio
from pydantic import BaseModel

class GuardrailResult(BaseModel):
    check: str
    passed: bool
    reason: str | None = None

async def check_jailbreak(query: str) -> GuardrailResult:
    # placeholder: swap in a classifier or your own heuristic model call
    await asyncio.sleep(0.05)
    return GuardrailResult(check="jailbreak", passed=True)

async def check_pii(query: str) -> GuardrailResult:
    await asyncio.sleep(0.05)
    return GuardrailResult(check="pii", passed=True)

async def check_topic(query: str) -> GuardrailResult:
    await asyncio.sleep(0.05)
    return GuardrailResult(check="topic", passed=True)

async def run_input_guardrails(query: str) -> list[GuardrailResult]:
    checks = [check_jailbreak(query), check_pii(query), check_topic(query)]
    return await asyncio.gather(*checks)  # parallel, not serial
```
